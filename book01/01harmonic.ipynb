```python
# This is a comment. It is not really part of the code. 
# Comments just provide helpful background information.

# This loads the library numpy and gives it the nickname "np"
import numpy as np

# This does similar for matplotlib. It's a library in python that
# generates MATLAB-like plots.
import matplotlib.pyplot as plt

# Set the minimum and maximum time value for my analysis
tmin = 0
tmax = 6 * np.pi

# Set the number of time points used for the analysis
Nt = 251

# This actually generates the points in time
t = np.linspace(tmin,tmax,Nt)

# y = cos(t)
y = np.cos(t)

# Plot it
plt.plot(t,y)

# Give the plot a proper title
plt.title('Fake Homework Problem 3.14: Displacement of An Undamped Spring')

# Put some proper axes labels on this plot
plt.xlabel('t [s]')
plt.ylabel('y [m]')

# This shows the plot.
plt.show()
```
